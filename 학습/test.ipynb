## 1. 혼자 놀기의 달인 (Lv.2 / 그래프)

```python
def 혼자놀기의달인(cards):
    cards = [-1] + cards
    v = [-1 for _ in range(len(cards))]
    groups = []

    def dfs(current):
        if v[current] != -1:
            groups.append(g)
            return
        
        g.append(cards[current])
        next = cards[current]
        v[current] = 1
        dfs(next)

    for i in range(1, len(cards)):
        if v[i] == -1:
            g = []
            dfs(i)
    
    groups.sort(key=lambda x:len(x))

    if len(groups) == 1: return 0
    return len(groups[-1]) * len(groups[-2])

test("예시1", 혼자놀기의달인([8,6,3,7,2,5,1,4]), 12)
test("예시2", 혼자놀기의달인([1,2,3,4,5]), 1)
test("예시3", 혼자놀기의달인([1]), 0)
```
